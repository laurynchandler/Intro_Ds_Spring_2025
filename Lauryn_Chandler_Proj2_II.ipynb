I used a histogram to look at the age distribution among the students in the dataset. It was interesting the data was clustered around 10-35 yet went all the way to 60. Therefore, I decided to look at the maximum and minimum to determine the age range, which was 18-59. To learn more about the distribution, I split the data into two age groups, 18-35 and 35-60, which are represented below.

```python
# Create subplots (2 rows, 2 columns)
fig, axes = plt.subplots(1, 2, figsize=(15, 8))

# Filter ages between 18 and 30
filtered_df1 = df_new[df_new['Age'].between(18, 35)]

# Histrogram with age range 18-35
sns.histplot(filtered_df1['Age'], bins=20, kde=True, ax=axes[0])
axes[0].set_title('Age Distribution (18–35)')
axes[0].set_xlabel('Age')
axes[0].set_ylabel('Frequency')


# Histrogram with age range 35-60
filtered_df2 = df_new[df_new['Age'].between(35, 60)]
sns.histplot(filtered_df2['Age'], bins=20, kde=True, ax=axes[1])
axes[1].set_title('Age Distribution (35-60)')
axes[1].set_xlabel('Age')
axes[1].set_ylabel('Frequency')

plt.show()
```

After breaking down the age groups, I had a better understanding of the age distribution. Most students were in the typical range for post-secondary school, and the average for the whole data set was 26 years old. Looking at the frequency comparisons, the 18-35 had thousands more than the 35-60. The outliers with ages up to 59 were interesting to me, so I decided to look a little more into them.

#### 1: Degree

```python
# Filter for age 35–60
filtered_df = df[(df['Age'] >= 35) & (df['Age'] <= 60)]

# Count number of students in each degree category
degree_counts = filtered_df['Degree'].value_counts().reset_index()
degree_counts.columns = ['Degree', 'Count']

# Bar plot
plt.figure(figsize=(10, 6))
sns.barplot(data=degree_counts, x='Degree', y='Count', hue='Degree', palette='pastel', legend=False)

plt.title('Number of Students Aged 35–60 by Degree')
plt.xlabel('Degree')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('students_35to60_by_degree.jpg')
plt.show()
```

I was interested in why the age range went so high, so I used a plot to look into what degrees students in the 35-60 age range were pursuing. I had assumed they would mostly go for advanced degrees such as medicine or PhDs, but the data showed differently. A high number were working towards a high school diploma, which could also be a GED program, as well as bachelor's degrees. It was interesting to see so many nontraditional students, possibly going back to get their education, which is inspiring.

#### 2: Average Work/Study Hours by Depression and Suicidal Thoughts

```python
grouped = df.groupby(['Depression', 'Have you ever had suicidal thoughts ?']).mean(numeric_only=True).reset_index()

# Bar plot of average Work/Study Hours
plt.figure(figsize=(10, 6))
sns.barplot(
    data=grouped,
    x='Have you ever had suicidal thoughts ?',
    y='Work/Study Hours',  
    hue='Depression'
)

plt.title('Average Work/Study Hours by Depression and Suicidal Thoughts')
plt.xlabel('Suicidal Thoughts')
plt.ylabel('Average Work/Study Hours')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('avgwork-studybydepression.jpg')
plt.show()
```

With this visualization, I decided to group data by depression and having suicidal thoughts. Then it was compared to average work/study hours. I assumed that higher averages would be common in people who had depression, suicidal thoughts, or both. Surprisingly, the data showed little difference between the two groups. People that did have depression did have higher average work/study hours, but they were quite equal whether having suicidal thoughts or not. The same was true for people without depression; the average work/study hours were about the same if suicidal thoughts were present or not. 

### Section 2: Relationship Visualization

```python
selected_cols = ['Depression', 'Financial Stress', 'Academic Pressure', 'Work Pressure', 'Work/Study Hours', 'Study Satisfaction', 'Job Satisfaction', 'CGPA', 'Age']
df_selected = df[selected_cols].copy()

# Convert Yes/No to 1/0
yes_no_cols = ['Depression']
df_selected[yes_no_cols] = df_selected[yes_no_cols].replace({'Yes': 1, 'No': 0})

# Heatmap looking at the correlation of selected variables
plt.figure(figsize=(8,6))
sns.heatmap(df_selected.corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Heatmap (Selected Variables)", fontsize=16)
plt.tight_layout()
plt.savefig('corrheatmap.jpg')
plt.show()
```

As I previously thought that different factors such as satisfaction, stress, and pressure would impact depression, I used a heatmap to look at the correlation between those variables. None of the variables had strong correlations with depression, which surprised me. The highest correlation was moderate and positive between academic pressure and depression. The heatmap only had one strong correlation, which was between job satisfaction and work pressure. While logical, it does not provide much insight into mental health.
