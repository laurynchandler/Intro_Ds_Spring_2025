```python
#have to import numpy and pandas in each notebook
import numpy as np
import pandas as pd
```

## **Task 1: NumPy Basics**

### **NumPy Array Creation**
* Create a 1D NumPy array of integers from 1 to 20.
* Create a 2D NumPy array (5x4) filled with random numbers.
* Display both arrays.

```python
#creates 1D array, () insures numbers 1 to 20 that are integers
arr = np.arange(1,21)
#creates 2D array and randomizes numbers, 5,4 sets the shape
arr2 = np.random.rand(5, 4)
print(arr)
print(arr2)
```

### **Array Indexing and Slicing**
* Extract all even numbers from the 1D array.
* Extract the second and fourth row from the 2D array

```python
even_numbers = arr[arr % 2 == 0]
#sets an array with just even numbers then prints
print(even_numbers)
```

```python
#extracts rows 1 and 3 out of the 5 and prints the whole row all columns
print(arr2[1, :]) 
print(arr2[3, :]) 
```

### **Mathematical Operations**
* Create a NumPy array of numbers from 10 to 100 with a step of 10.
* Perform element-wise square and square root operations on this array

```python
#array from 10 to 100 with steps of 10
a = np.arange(10, 101,10)
#makes element-wise square and sqaure root form given array
squared_a = np.square(a)
sqrt_a = np.sqrt(a) 
print("Square:", squared_a)
print("Square Roots:", sqrt_a)
```

### **Statistical Operations**
* Generate a 3x3 NumPy array of random integers between 1 and 50.
* Compute the mean, median, and standard deviation.

```python
#3 by 3 array with integers from 1 to 50
b = np.random.randint (1,50,(3, 3))
print(b)
#gives the descriptive statistics
print('Mean:', np.mean(b))
print('Median:', np.median(b))
print('Standard Deviation:', np.std(b))
```

## **Task 2: Pandas Basics**

### **Data Loading and Overview**
* Load the dataset using Pandas.
* Display the first five rows.
* Show dataset information and check for missing values.
* Copy the original data to another data frame, df_new before doing any activities below
(sec 2.2 onwards).

```python
#calling the file from notebook and printing
df = pd.read_csv("project1_part2_data.csv")
#head prints first 5 rows
print(df.head())
#shows dataset info
print(df.info()) 
#checks for missing values
print(df.isnull().sum())
#copy dstaset
df_new = df.copy()
```

### **Data Selection and Filtering**
* Extract only the Name and Salary columns.
* Filter employees who earn more than $50,000

```python
#extract name and salary columns
df_extracted = df_new[['Name','Salary']]
#then shows name and salary of employees who make over 50,000
df_extracted.loc[df_extracted['Salary'] > 50000.0]
```

### **Data Aggregation**
* Find the average salary of employees.
* Count the number of employees in each department.

```python
#average salary of employees
average_salary = df_new['Salary'].mean()
print(average_salary)
# group by department and count the number of employees in each
department_grouped=df_new.groupby('Department').count()
print(department_grouped)
```

### **Data Cleaning**
* Replace any missing Salary values with the mean salary.
* Drop rows where Age is missing.

```python

# Replace missing salary with average salary 
df_new['Salary']=df_new['Salary'].fillna(df_new['Salary'].mean()) 
# drop missing age
df_cleaned = df_new.dropna(subset=['Age'])
df_cleaned
```

```python
#check for missing values again
df_cleaned.isnull().sum()
```

## **Task 3: Combined NumPy and Pandas Operations**

### **Converting a Column to a NumPy Array**
* Extract the Salary column as a NumPy array.
* Compute the standard deviation using NumPy.

```python
#extract salary column as array
salary_array = df_cleaned['Salary'].to_numpy()
print(salary_array)
#standard deviation of array
print('Standard Deviation:', np.std(salary_array))
```

### **New Column Creation**
* Create a new column Salary_Bonus where each employee gets a 10% increase in salary

```python
#new column with bonus
df_cleaned['Salary_Bonus'] = df_cleaned['Salary']*1.10
print(df_cleaned)
```

### **Grouping, Summarization and Saving**
* Group data by Department and find the total salary expense per department.
* Save the cleaned csv as project1_part2_data_cleaned.csv

```python
#group by departnment, then total salary expense per department
expense = df_cleaned.groupby('Department')['Salary'].sum()
print(expense)
```

```python
#save cleaned data
df_cleaned.to_csv('project1_part2_data_cleaned.csv', index=False)
```
