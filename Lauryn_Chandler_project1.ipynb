#### Q1<br>
The different data types in Python are: <br>
* int: integer, used for whole numbers ex: 7
* float: used for decimals ex: 4.3
* string: used for a string of data ex: str = 'Lauryn'
* boolean: used for true or false bool: true
* list: used to store data in one variable ex: list1 = ['red', 'blue', 'green']
* tuple: used to store data ex: tuple1 = ('red', 'blue', 'green') 
* set: holds a set of variables ex: set = (red, blue, green)
* dictionary: holds and stores keys and their values ex: dict1= {'color': 'red'}
