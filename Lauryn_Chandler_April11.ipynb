### Section 1: Dataset Overview and Structure

```python
# Import the Libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Load the dataset
df = pd.read_csv('Student Depression Dataset.csv')
df
```

```python
#create a copy of the data so I can replace the missing values
df_new=df.copy()
df_new
```

```python
df_new['Financial Stress']=df_new['Financial Stress'].fillna(df_new['Financial Stress'].mean())  # Replace missing Financial Stress scores with mean
df_new
```

Lower sleep duration tends to have higher work/study hours.

```python
df_new.to_csv('Student Depression Dataset_cleaned.csv')
#saving the cleaned data
```

```python
selected_cols = ['Depression', 'Financial Stress', 'Academic Pressure', 'Work Pressure', 'Work/Study Hours', 'Study Satisfaction', 'Job Satisfaction', 'CGPA', 'Age']
df_selected = df[selected_cols].copy()

# Convert Yes/No to 1/0
yes_no_cols = ['Depression']
df_selected[yes_no_cols] = df_selected[yes_no_cols].replace({'Yes': 1, 'No': 0})

# Heatmap
plt.figure(figsize=(8,6))
sns.heatmap(df_selected.corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Heatmap (Selected Variables)", fontsize=16)
plt.tight_layout()
plt.savefig('corrheatmap.jpg')
plt.show()
```

```python
plt.hist(df_new['Age'],bins=30, color='skyblue', edgecolor='black')

plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.savefig('agedistribution.jpg')
plt.show()
```
